## Exploratory Data Analysis (EDA)

### Objective

The objective of this exploratory data analysis is to understand how demographic and lifestyle factors influence medical insurance charges. This analysis explores variables such as age, sex, BMI, number of children, smoking status, and region to identify trends, patterns, and relationships that affect healthcare costs.

```python
# import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load datasets
insurance = pd.read_csv('insurance.csv')
```

```python
# display the first five rows
insurance.head()
```

```python
# check the shape of the dataset
insurance.shape
```

```python
# check information about the dataset
insurance.info()
```

```python
# summary statistics

insurance.describe()
```

```python
# check for missing values

insurance.isnull().sum()
```

```python
# check for duplicate records

insurance.duplicated().sum()
```

```python
# Remove duplicate records
insurance.drop_duplicates(inplace=True)
```

```python
# confirm the new shape
insurance.shape
```

```python
insurance.describe()
```

## Summary Statistics

The summary statistics provide an overview of the numerical variables in the dataset, including age, BMI, number of children, and medical charges. They help identify the range, average, and spread of each variable before further analysis.

```python
plt.figure(figsize=(8,5))
plt.hist(insurance['age'], bins=15, edgecolor='black')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

## Distribution of Age

This histogram shows the distribution of customers' ages in the dataset. It helps identify whether the data is concentrated around certain age groups or evenly distributed.

## Distribution of BMI

This histogram illustrates the distribution of Body Mass Index (BMI) among customers, helping to understand the overall health profile of the insured individuals.

```python
plt.figure(figsize=(8,5))
plt.hist(insurance['bmi'], bins=15, edgecolor='black')
plt.title('Distribution of BMI')
plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.show()
```

## Distribution of Medical Charges

This histogram displays the distribution of insurance charges. It helps identify whether medical costs are evenly distributed or concentrated among a smaller group of customers.

```python
plt.figure(figsize=(8,5))
plt.hist(insurance['charges'], bins=20, edgecolor='black')
plt.title('Distribution of Medical Charges')
plt.xlabel('Medical Charges')
plt.ylabel('Frequency')
plt.show()
```

## Smoking Status Distribution

This chart shows the number of smokers and non-smokers in the dataset.

```python
insurance['smoker'].value_counts().plot(kind='bar')
plt.title('Smoking Statis')
plt.xlabel('Smoker')
plt.ylabel('Count')
plt.show()
```

## Gender Distribution

This chart displays the number of male and female customers in the dataset.

```python
insurance['sex'].value_counts().plot(kind='bar')

plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

## Region Distribution

This chart shows how customers are distributed across different regions.

```python
insurance['region'].value_counts().plot(kind='bar')

plt.title('Region Distribution')
plt.xlabel('Region')
plt.ylabel('Count')
plt.show()
```

## Distribution of Number of Children

This chart shows the number of dependents covered by the insurance policy.

```python
insurance['children'].value_counts().sort_index().plot(kind='bar')
plt.title('Number of Children')
plt.xlabel('Children')
plt.ylabel('Counts')
plt.show()
```

## Age vs Medical Charges

This scatter plot examines the relationship between age and medical insurance charges.

```python
plt.figure(figsize=(8,5))
plt.scatter(insurance['age'], insurance['charges'])

plt.title('Age vs Medical Charges')
plt.xlabel('Age')
plt.ylabel('Medical charges')
plt.show()
```

### Insight

Medical charges generally increase with age. Older individuals tend to incur higher healthcare costs than younger individuals.

## BMI vs Medical Charges

This scatter plot shows the relationship between Body Mass Index (BMI) and medical insurance charges.

```python
plt.figure(figsize=(8,5))
plt.scatter(insurance['bmi'],insurance['charges'])

plt.title('BMI vs Medical charges')
plt.xlabel('BMI')
plt.ylabel('Medical Charges')
plt.show()
```

### Insight

Customers with higher BMI values tend to have higher medical charges, although the relationship is not perfectly linear.

## Smoking Status vs Medical Charges

This box plot compares insurance charges between smokers and non-smokers.

```python
plt.figure(figsize=(8,5))
insurance.boxplot(column='charges',by='smoker')

plt.title('Charges by Smoking Status')
plt.suptitle('')
plt.xlabel('Smoker')
plt.ylabel('Medical Charges')
plt.show()
```

### Insight

Smokers have significantly higher medical charges than non-smokers, indicating that smoking is a major factor affecting healthcare costs.

## Gender vs Medical Charges

This box plot compares medical insurance charges between male and female customers.

```python
plt.figure(figsize=(8,5))

insurance.boxplot(column='charges',by='sex')

plt.title('Charges by Gender')
plt.suptitle('')
plt.xlabel('Gender')
plt.ylabel('Medical Charges')
plt.show()
```

### Insight

Male and female customers have comparable medical insurance charges. Gender alone does not appear to have a strong influence on healthcare costs.

## Region vs Medical Charges

This box plot compares medical insurance charges across different geographical regions.

```python
plt.figure(figsize=(8,5))

insurance.boxplot(column='charges',by='region')

plt.title('Charges by Region')
plt.suptitle('')
plt.xlabel('Region')
plt.ylabel('Medical charges')
plt.show()
```

### Insight

Medical charges are fairly similar across the regions, with only slight variations. Region appears to have a relatively small impact on insurance charges.

## Number of Children vs Medical Charges

This box plot examines how the number of children covered by the insurance policy relates to medical charges.

```python
plt.figure(figsize=(8,5))

insurance.boxplot(column='charges',by='children')

plt.title('Charges by Number of Children')
plt.suptitle('')
plt.xlabel('Number of Children')
plt.ylabel('Medical Charges')
plt.show()
```

### Insight

The number of children does not show a strong relationship with medical insurance charges. Charges are distributed similarly across different family sizes.

## Correlation Heatmap

The heatmap shows the correlation between numerical variables in the dataset.

```python
numeric_data = insurance.select_dtypes(include=['number'])

corr = numeric_data.corr()

plt.figure(figsize=(8,6))
plt.imshow(corr, cmap='coolwarm')

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title('Correlation Heatmap')
plt.show()
```

### Insight

The heatmap indicates that **charges** have the strongest positive correlation with **age** and a moderate positive correlation with **BMI**. Other numerical variables show relatively weak relationships with insurance charges.

# Exploratory Data Analysis Summary

The exploratory data analysis identified several important patterns in the medical insurance dataset.

- Medical insurance charges generally increase with age.
- Individuals with higher BMI tend to have higher healthcare costs.
- Smoking has the strongest impact on insurance charges, with smokers paying substantially more than non-smokers.
- Gender and region have relatively small effects on medical charges.
- The number of children covered under the insurance policy does not significantly influence medical expenses.

Overall, smoking status, age, and BMI appear to be the most influential factors affecting healthcare costs. These findings provide useful insights for developing a regression model to predict medical insurance charges.
